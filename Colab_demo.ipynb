```
import os
import subprocess

# Specify the top-level directory where your nested folders are
base_dir = "/content/drive/MyDrive/short_mp4"

# Create a directory for RIFE output
output_dir = "/content/drive/MyDrive/rife_videos"
os.makedirs(output_dir, exist_ok=True)

# Loop through all directories and files inside the base directory
for root, dirs, files in os.walk(base_dir):
    for file in files:
        if file.endswith(".mp4"):
            video_path = os.path.join(root, file)
            print(f"Processing {video_path}...")

            # Generate the output video path
            output_video_path = os.path.join(output_dir, f"rife_{file}")

            # Run RIFE interpolation without comparison
            command = f"python3 /content/arXiv2020-RIFE/inference_video.py --video={video_path} --output={output_video_path}"

            # Run the RIFE command
            subprocess.run(command, shell=True)

            print(f"Finished processing: {output_video_path}")
```

```
import os
import subprocess

# Specify the directory where your RIFE videos are located
input_dir = "/content/drive/MyDrive/rife_videos"

# Create a directory for the slowed-down videos
output_dir = "/content/drive/MyDrive/rife_videos_slow"
os.makedirs(output_dir, exist_ok=True)

# Loop through all video files in the rife_videos folder
for root, dirs, files in os.walk(input_dir):
    for file in files:
        if file.endswith(".mp4"):
            video_path = os.path.join(root, file)
            print(f"Slowing down: {video_path}...")

            # Generate the output video path with '_slow' suffix
            output_video_path = os.path.join(output_dir, f"slow_{file}")

            # FFmpeg command to slow down video by 0.2x
            command = f"ffmpeg -i {video_path} -filter:v \"setpts=5.0*PTS\" -an {output_video_path}"

            # Run the command
            subprocess.run(command, shell=True)

            print(f"Finished slowing down: {output_video_path}")
```
